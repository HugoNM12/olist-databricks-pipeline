```python
import uuid
from datetime import datetime
from pyspark.sql.types import (StructType, StructField, StringType,
                               TimestampType, LongType, DoubleType)

def criar_tabela_audit():
    spark.sql("CREATE SCHEMA IF NOT EXISTS workspace.audit")
    spark.sql(f"""
        CREATE TABLE IF NOT EXISTS {TABELA_AUDIT} (
            run_id STRING, camada STRING, tabela STRING,
            inicio TIMESTAMP, fim TIMESTAMP, duracao_seg DOUBLE,
            status STRING, linhas_esperadas BIGINT,
            linhas_gravadas BIGINT, mensagem_erro STRING)
    """)

def record_audit(registros):
    schema = StructType([
        StructField("run_id", StringType()),
        StructField("camada", StringType()),
        StructField("tabela", StringType()),
        StructField("inicio", TimestampType()),
        StructField("fim", TimestampType()),
        StructField("duracao_seg", DoubleType()),
        StructField("status", StringType()),
        StructField("linhas_esperadas", LongType()),
        StructField("linhas_gravadas", LongType()),
        StructField("mensagem_erro", StringType()),
    ])
    df = spark.createDataFrame(registros, schema)
    df.write.mode("append").saveAsTable(TABELA_AUDIT)
```
